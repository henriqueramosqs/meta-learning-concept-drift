```python
rename_map ={
    "proposed_mtl": "proposed_mfes",
    "original_mtl": "original_mfes"
}

COLORS = [
   
    "#cce1e4", # green
    # "#d9d9d9", # grey
     "#ffd7b3", # orange
    "#FF6347" # red
    ]

def plot_mean_mse(group_by='dataset'):
    grouped_df = mean_results_df.groupby([group_by]).mean(True)
    grouped_df = grouped_df.rename(columns={"proposed_mtl": 'fernanda\'s mtl\'s'})

    grouped_df_h = mean_results_df_h.groupby([group_by]).mean(True)   
    grouped_df = grouped_df.join(grouped_df_h['proposed_mtl'])
    grouped_df = grouped_df.rename(columns=rename_map)
    cols_2_plot = ["baseline", "original_mfes","proposed_mfes"]
    print(grouped_df,"\n")
    grouped_df[cols_2_plot].plot.bar(figsize=(15,15), fontsize=16, color=COLORS, rot=0).legend(loc='best', prop={'size': 16})

print(mean_results_df.columns)
plot_mean_mse(group_by="dataset")
plot_mean_mse(group_by="model")
plot_mean_mse(group_by="metric")
```

```python
BASE_MODELS = [
    "RandomForestClassifier",
    "SVC",
    "LogisticRegression",
    "DecisionTreeClassifier"
]

def get_final_gain(base_model,eval,plot_col,metric="recall"):
    results_df = eval.results[base_model]
    if plot_col == "ideal_regressor":
        results_df[f"{metric}_mse_{plot_col}"] = 0
    regressor_error = results_df[f"{metric}_mse_{plot_col}"]
    baseline_error = results_df[f"{metric}_mse_baseline"]
    proposed_mtl_gain = baseline_error - regressor_error
    y = proposed_mtl_gain.cumsum()
    return list(y)[-1]



aux = []
for evaluator in (
                  elec_eval_h, 
                  powersupply_eval_h, 
                  rialto_eval_h,
                  airlines_eval_h, 
                  ):
    for base_model in BASE_MODELS:
        for metric in ["recall", "f1-score", "kappa", "precision"]:
            print(f"metric: {metric}")
            original = get_final_gain(base_model, evaluator,"without_drift",metric)
            proposed = get_final_gain(base_model, evaluator,"with_drift",metric)
            ideal = get_final_gain(base_model, evaluator,"ideal_regressor",metric)
            print(f"BASE_MODEL: {base_model} -> original: {original/ideal}, proposed: {proposed/ideal}")
            aux.append(((proposed/ideal)-(original/ideal), f"{metric}, {base_model}, {evaluator.dataset_name}" ))
    print("="*50)
    aux.sort(reverse=True)
    print(aux)
```
