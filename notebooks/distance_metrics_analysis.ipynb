```python
def open(notebook_dir:str, filename)-> pd.DataFrame:
    script_dir = os.getcwd()
    print(script_dir)
    path = os.path.abspath(os.path.join(script_dir, '../', 'metabase', notebook_dir, filename))
    # notebook_path = os.path.join(os.path.abspath('../metabase', notebook_dir))  
    print("path", path)
    

    if not os.path.exists(path):
        print(f"ERRO: Arquivo não encontrado em {path}")

    df = pd.read_csv(path)
    return df
```
